# Demo bundle for the API and Streamlit app

Inputs: the **week 8** notebook's output and the **week 8b** (ONNX) notebook's output. Accelerator **None**, Internet **on**.
Output: `demo_bundle.zip` (about 5 MB): unzip it into the repo's `demo/` folder.

```python
import os, sys, subprocess, shutil
from pathlib import Path
REPO = "https://github.com/rc-mathew/deep-learning-fraud-detection.git"
if not os.path.exists("deep-learning-fraud-detection"):
    subprocess.run(["git", "clone", "-q", REPO], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "captum", "faiss-cpu", "onnxruntime"], check=True)
sys.path.insert(0, "deep-learning-fraud-detection/src")
ART = Path("/kaggle/working/artifacts")
w8 = next(Path("/kaggle/input").rglob("week7_scores.npz")).parent
onnx = next(Path("/kaggle/input").rglob("transformer_fp32.onnx")).parent
for name in ("sequences", "models"):
    shutil.copytree(w8 / name, ART / name, dirs_exist_ok=True)
for name in ("week7_scores.npz", "transactions.parquet"):
    shutil.copy(w8 / name, ART / name)
shutil.copytree(onnx, ART / "onnx", dirs_exist_ok=True)
```

```python
from fraud_dl.build_demo import build
out = build(ART, Path("/kaggle/working/demo"))
shutil.make_archive("/kaggle/working/demo_bundle", "zip", out)
shutil.rmtree(ART)  # keep the output small: only the bundle is needed
print(sorted(p.name for p in out.iterdir()), os.path.getsize("/kaggle/working/demo_bundle.zip") / 1e6, "MB")
```
